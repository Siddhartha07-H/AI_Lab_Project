**Create a semantic network with visualization using Python.**

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 10))

def circle(x, y, text):
    plt.scatter(x, y, s=5000, color="lightblue")
    plt.text(x, y, text, ha="center", va="center",
             fontsize=12, fontweight="bold")

def arrow(x1, y1, x2, y2, text, lx, ly):
    plt.annotate("", (x2, y2), (x1, y1),
                 arrowprops=dict(arrowstyle="->", linewidth=2))
    plt.text(lx, ly, text, ha="center", fontsize=11)

circle(5, 10, "SIDDHARTHA")
circle(5, 8, "STUDENT")
circle(5, 6, "AI")
circle(2, 4, "DATA MINING")
circle(8, 4, "MACHINE\nLEARNING")
circle(5, 2, "PYTHON")

arrow(5, 9.5, 5, 8.5, "is a", 5.4, 9)
arrow(5, 7.5, 5, 6.5, "studies", 5.5, 7)
arrow(4.5, 5.5, 2.5, 4.5, "includes", 3, 5)
arrow(5.5, 5.5, 7.5, 4.5, "includes", 7, 5)
arrow(2.5, 3.5, 4.5, 2.5, "uses", 3.2, 3)
arrow(7.5, 3.5, 5.5, 2.5, "uses", 6.8, 3)

plt.xlim(0, 10)
plt.ylim(0, 11)
plt.axis("off")
plt.show()
```
